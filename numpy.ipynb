Aliasing the arrays - Aliasing is not copying .It means giving another name to the existiong object and vice versa.
syntax--b=a
--->In Aliasing, memory is not located to array b
--->a and b both reference to the object i.e array

```python
a=arange(1,6)
b=a
print("ORiginal arr: ",a)
print("Aliasing array: ",b)
b[0]=99
print("Now,Original arr: ",a)
print("Aliasing arr",b)
```

3. View() and copy() function:
view(): Itis used for "shallow copying".
        Syntax--b=a.view()
        The memory is allocated to the array b
        The modifications done in array b will also modified array a.


copy(): It is used for "deep copying".
        when modifications done on array b doesnot affect array a.

```python
#View() Function
a=arange(1,5)
b=a.view()
b[0]=99
print(a,b,sep="       ")
#Copy() Function
p=arange(5,10)
q=p.copy()
q[0]=56
print(p,q,sep="     ")
```

```python
#2D array
arr=[[1,2,3],[5,6,7],[8,9,3]]
#display only rows
for i in range(len(arr)):
    print(f"row:{i} and elements are: {arr[i]}")
#display only columns
for i in range(len(arr)):
    for j in range(len(arr[i])):
        print(arr[i][j], end=" ")
```

## Handling missing data
1.np.isnan()===detect missing values--always return boolean value(true/false)
# we can not compare two nan values
2.np.nan_to_num()===not a number to number conversion
3.no.isinf===detect infinity

```python
import numpy as np
arr=np.array([1,nan,3,nan,5])
print(np.isnan(arr))
#print(np.nan_to_num(arr,nan=100))
print(np.nan_to_num(arr))
```

```python
import numpy as np
arr=np.array([1,np.inf,3,-np.inf,5])
print(np.isinf(arr))
cleaned_arr=np.nan_to_num(arr,posinf=100,neginf=-100)
print(cleaned_arr)
```
